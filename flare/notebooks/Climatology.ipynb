```python
import pandas as pd

goes_fl=pd.read_csv('/home/baydin2/workspace/flarepredictiondata/flare_reading/datain/final/goes_aia_validated.csv',
                     parse_dates=['event_date','start_time', 'peak_time', 'end_time'])
goes_fl.set_index('flare_id')
```

```python
start_date = min(goes_fl.event_date)
end_date = max(goes_fl.event_date)
tranges = pd.date_range(start=start_date, end=end_date,freq='24h')
#tranges
```

```python

def get_climatology(fldf, start, end):
    tranges = pd.date_range(start=start_date, end=end_date,freq='24h')

    count = 0
    for i, trs in enumerate(tranges):
        if i+1 == len(tranges):
            break
        tre = tranges[i+1]
        tdf = fldf[(trs <= fldf['peak_time']) & (fldf['peak_time'] < tre)]
        if tdf.shape[0] > 0:
            count = count + 1
    
    return float(count) / len(tranges)
            
xf = goes_fl[goes_fl['goes_class'].str.startswith('X')]
mf = goes_fl[goes_fl['goes_class'].str.startswith('M')]
cf = goes_fl[goes_fl['goes_class'].str.startswith('C')]
bf = goes_fl[goes_fl['goes_class'].str.startswith('B')]

mx = goes_fl[goes_fl['goes_class'] > 'M']
bc = goes_fl[goes_fl['goes_class'] < 'M']

start_date = min(goes_fl.event_date)
end_date = max(goes_fl.event_date)


climatology = {'x':get_climatology(xf,start_date, end_date),
               'm':get_climatology(mf,start_date, end_date),
               'c':get_climatology(cf,start_date, end_date),
               'b':get_climatology(bf,start_date, end_date),
               'mx':get_climatology(mx,start_date, end_date),
               'bc':get_climatology(bc,start_date, end_date),
               }
climatology
```
